# KE Recovery - chạy toàn bộ thí nghiệm một lượt
Bấm **Thời gian chạy -> Chạy tất cả**. Các bước (S0, S1, ...) chạy theo thứ tự, dừng ngay nếu một cổng kiểm tra không qua.
Bước đã có `results/<bước>/DONE.json` được bỏ qua; đặt `FORCE = True` để chạy lại tất cả, hoặc `ONLY = ['S1']` để chạy riêng một bước.

```python
from google.colab import drive
drive.mount('/content/drive')
ROOT = '/content/drive/MyDrive/KE_Recovery'
FORCE = False   # True: chạy lại mọi bước, kể cả bước đã xong
ONLY = None     # ví dụ ['S1'] để chạy riêng một bước
ALLOW_TEST_RERUN = False  # True: cho phép chạy lại tập test dù đã có TEST_LOCK.json (việc chạy lại sẽ bị ghi nhận)
import sys, importlib
sys.path.insert(0, f'{ROOT}/src')
importlib.invalidate_caches()   # Python lưu đệm danh sách file: làm mới để thấy file mới trên Drive
for _m in [m for m in sys.modules if m.split('.')[0] in ('ke_research', 'backend', 'strategies')]:
    del sys.modules[_m]                    # bỏ bản module cũ còn trong bộ nhớ phiên
from ke_research import pipeline
```

```python
log = pipeline.run_all(ROOT, only=ONLY, force=FORCE, allow_test_rerun=ALLOW_TEST_RERUN)
failed = [e for e in log if e['status'] == 'FAILED']
assert not failed, f"Pipeline dừng ở {failed[0]['step']}: {failed[0]['error']}"
```
